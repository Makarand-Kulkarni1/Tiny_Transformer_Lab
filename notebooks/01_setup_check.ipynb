```python
import sqlite3

with open("../src/data/schema.sql") as f:
    schema = f.read()

conn = sqlite3.connect(":memory:")   # a temporary database that lives only in RAM
conn.executescript(schema)           # runs all three CREATE TABLE statements

tables = conn.execute(
    "SELECT name FROM sqlite_master WHERE type='table'"
).fetchall()
print(tables)
```

```python
conn.execute("INSERT INTO customers VALUES (1, 'asha', 'pune')")
conn.execute("INSERT INTO products VALUES (1, 'notebook', 'stationery', 49.5)")
conn.execute("INSERT INTO orders VALUES (1, 1, 1, 3, '2026-03-15')")

result = conn.execute("""
    SELECT customers.name, products.name, orders.quantity
    FROM orders
    JOIN customers ON orders.customer_id = customers.id
    JOIN products  ON orders.product_id  = products.id
""").fetchall()
print(result)
```
